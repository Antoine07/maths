```python
# population total 
x = np.array([1, 2, 3])
y = np.array([2,4,5])

# Matrice de covariance 
matrix_cov = np.cov(x, y)
matrix_cov
```

```python
#  Var(x) Cov(x,y)
#  Cov(y,x) Var(y)

# Cov(x,y) = Cov(y,x)

cov = matrix_cov[0,1]

cov
```

```python
std_x = np.std(x, ddof=1) # degré de liberté 1 on prend la pop
std_y = np.std(y, ddof=1)

rho = cov / (std_x * std_y)
```

```python
rho
```
